## 2. Okruženje i pomoćne funkcije

Ne treba ništa instalirati: `pandas`, `numpy`, `matplotlib`, `scipy` i `scikit-learn` već postoje u Colabu. Jedino što ćemo dodati kasnije je `google-genai`, paket za poziv jezičnog modela.

U ovoj ćeliji postavljamo i **izgled slika** (tamna pozadina, boje kolegija) i dvije male pomoćne funkcije koje će nam štedjeti tipkanje:

- `nadji_stupac(df, ...)` — pronalazi stupac po dijelu imena. Zašto? Zato što Google obrasci imena stupaca prepisuju **doslovno** iz pitanja, pa se razmak, veliko slovo ili nova riječ u pitanju odmah preslikaju u ime stupca. Kod koji se veže za točno ime pukne čim predavač popravi tipfeler u pitanju.
- `udio(vrijednosti)` — pretvara brojeve u postotke za ispis uz sliku.


```python
import io, json, os, random, re, time, urllib.request
from collections import Counter

import numpy as np
import pandas as pd

print('pandas', pd.__version__, '| numpy', np.__version__, '| matplotlib', plt.matplotlib.__version__)


def nadji_stupac(df, *kljucevi):
    'Vraca ime prvog stupca koje sadrzi neki od zadanih kljuceva (bez obzira na velicinu slova).'
    for k in kljucevi:
        for c in df.columns:
            if k.lower() in str(c).lower():
                return c
    return None


def udio(n, ukupno):
    'Postotak za ispis uz broj.'
    return round(100 * n / ukupno, 1) if ukupno else 0.0


def tablica(serija, naslov_lijevo='Odgovor', naslov_desno='Broj', ukupno=None):
    'Ispisuje seriju kao mali pregled s postocima.'
    ukupno = ukupno or int(serija.sum())
    out = pd.DataFrame({naslov_lijevo: serija.index, naslov_desno: serija.values})
    out['%'] = [udio(v, ukupno) for v in serija.values]
    return out
```

## 3. Kako doći do podataka (tri puta)

Prije bilo kakve analize moramo odgovoriti na jedno pitanje: **odakle podaci ulaze u bilježnicu?** Postoje tri načina, od najboljeg prema rezervnom.

**Put A — objavljeni CSV iz Google tablice (preporučeno za prave podatke).**
U obrascu otvori karticu *Responses (Odgovori)* → *Link to Sheets*; nastane Google tablica s odgovorima. U toj tablici: *File → Share → Publish to web*, odaberi list s odgovorima i format **CSV**, klikni *Publish*. Dobiješ poveznicu koja izgleda ovako:

`https://docs.google.com/spreadsheets/d/DUGI_ID/pub?gid=0&single=true&output=csv`

Tu poveznicu upiši u varijablu `PRAVI_ODGOVORI_URL`. Ona se osvježava sama — svaki novi odgovor u obrascu pojavi se u tablici, a time i u ovoj bilježnici.

**Put B — ručni uvoz datoteke.** Ako ne želiš objavljivati tablicu, u tablici odaberi *File → Download → CSV*, pa datoteku prenesi u Colab (ćelija s `files.upload()`), ili je povuci na ikonu mape u lijevoj traci.

**Put C — pokazni podaci.** Ako pravih odgovora još nema (ili ih ima premalo za smislenu analizu), bilježnica uzima **pokazne odgovore** s repozitorija kolegija i sve radi na njima. Ti podaci su generirani programski, imaju istu strukturu kao pravi i služe isključivo za vježbu — na slici i u brojkama to uvijek piše.

> **O privatnosti.** Upitnik ne traži ime ni adresu. Ako ikada objavljuješ odgovore, objavi **zbirne** rezultate ili anonimiziraj tablicu: ime, adresa i slobodan tekst u kojem se netko predstavio ne idu u javnost. U ovoj bilježnici analiziramo isključivo anonimne odgovore.


```python
# ---------------------------------------------------------------------------
# 1) PRAVI podaci: ovdje upisi poveznicu na OBJAVLJENE odgovore (CSV iz tablice).
#    Prazno = koriste se pokazni podaci.
PRAVI_ODGOVORI_URL = ''

# 2) Pokazni podaci kolegija (isti upitnik, generirani odgovori za vjezbu).
POKAZNI_URL = 'https://raw.githubusercontent.com/bperak/dsk/main/data/demo_odgovori.csv'
print('Poveznica za prave odgovore:', PRAVI_ODGOVORI_URL or '(nije upisana — radimo na pokaznim podacima)')
```

```python
def dohvati_csv(url, vrijeme=20):
    'Dohvaca CSV s poveznice i vraca ga kao tekst ili None ako ne uspije.'
    try:
        zahtjev = urllib.request.Request(url, headers={'User-Agent': 'DSK-kolegij/1.0'})
        with urllib.request.urlopen(zahtjev, timeout=vrijeme) as odgovor:
            return odgovor.read().decode('utf-8', errors='replace')
    except Exception as greska:
        print('Nije uspjelo dohvacanje s', url[:70], '->', type(greska).__name__, str(greska)[:90])
        return None


def pokazni_ako_treba(broj=38, sjeme=20261013):
    'Generira pokazne odgovore ako nijedna poveznica ne radi — biljeznica mora raditi i offline.'
    rnd = random.Random(sjeme)
    mjesta = ['Kino Cinestar', 'Kino Art-kino', 'Hrvatsko narodno kazaliste "Ivan Zajc"',
              'Hrvatski kulturni dom "Susak"', 'Klub Nemo (ex. Zivot)', 'Klub Crkva',
              'Klub mladih', 'Kafe Dnevni boravak']
    tezine = [0.16, 0.42, 0.12, 0.18, 0.55, 0.48, 0.30, 0.62]
    glazba = ['Taylor Swift; The Beatles; Radiohead', 'the beatles ; radiohead',
              'Severina; Jelena Rozga; Thompson', 'AC/DC; Metallica; Iron Maiden',
              'Kendrick Lamar; Travis Scott', 'klasična glazba; Bach; Chopin',
              'Dubioza kolektiv; Zabranjeno pušenje', 'BTS; BLACKPINK; NewJeans']
    redovi = []
    for i in range(broj):
        odabir = [m for m, t in zip(mjesta, tezine) if rnd.random() < t]
        while len(odabir) < 2:
            odabir.append(rnd.choices(mjesta, weights=tezine, k=1)[0])
        redovi.append({
            'Gdje izlazite u Rijeci': ', '.join(dict.fromkeys(odabir)),
            'Dob': str(int(min(max(round(rnd.gauss(20.4, 1.7)), 18), 27))),
            'Koju glazbu slušate': glazba[i % len(glazba)],
            'Uživam slušati narodnjake koji se puštaju u kafićima u centru Rijeke': str(min(max(int(round(rnd.gauss(2.1, 1.0))), 1), 5)),
            'Slušam u slobodno vrijeme glazbu Jochana Sebastiana Bacha': str(min(max(int(round(rnd.gauss(2.6, 1.0))), 1), 5)),
        })
    return pd.DataFrame(redovi)


def ucitaj_podatke():
    'Vraca (podaci, izvor). Redoslijed: pravi CSV -> pokazni CSV s repozitorija -> lokalna datoteka -> generirani pokazni.'
    for url, izvor in [(PRAVI_ODGOVORI_URL, 'PRAVI odgovori iz Google tablice'),
                       (POKAZNI_URL, 'pokazni odgovori s repozitorija kolegija')]:
        if not url:
            continue
        tekst = dohvati_csv(url)
        if tekst:
            try:
                probni = pd.read_csv(io.StringIO(tekst))
            except Exception:
                probni = None
            if probni is not None and len(probni) >= 1 and probni.shape[1] >= 3:
                return probni, izvor
    for lokalno in ['demo_odgovori.csv', '/content/demo_odgovori.csv']:
        if os.path.exists(lokalno):
            return pd.read_csv(lokalno), 'pokazni odgovori iz datoteke ' + lokalno
    return pokazni_ako_treba(), 'generirani pokazni odgovori (nista nije dohvaceno)'


podaci, IZVOR = ucitaj_podatke()
print('IZVOR PODATAKA:', IZVOR)
print('Redaka:', len(podaci), '| stupaca:', podaci.shape[1])
print('Stupci:', list(podaci.columns))
podaci.head()
```

Uz podatke u pravom istraživanju uvijek ide i **broj ispitanika**. Trik je u tome što broj redaka nije nužno broj ljudi: netko može poslati obrazac dvaput, netko može preskočiti dio pitanja. Zato prije analize provjeravamo: koliko redaka imamo, koji stupci imaju praznine i koliko je zapisa nejasno.


## 4. Prvi pogled na tablicu: što uopće imamo?

Prvo čitanje podataka nije analiza — to je **upoznavanje**. Gledamo pet stvari: koliko je redaka i stupaca, kojeg su tipa, koliko ima praznina, koliko različitih vrijednosti ima svaki stupac i kako izgleda nekoliko prvih redaka.

Zašto baš to? Zato što su točno te četiri stvari izvor gotovo svake pogreške u studentskim radovima: broj se čita kao tekst, praznina se pretvori u nulu, a stupac s više odgovora u jednoj ćeliji analizira se kao da je jedan odgovor.


```python
pregled = pd.DataFrame({
    'tip': [str(t) for t in podaci.dtypes],
    'praznina': [int(podaci[c].isna().sum()) for c in podaci.columns],
    'različitih': [int(podaci[c].nunique(dropna=True)) for c in podaci.columns],
    'primjer': [str(podaci[c].dropna().iloc[0])[:48] if podaci[c].notna().any() else '(sve prazno)'
                for c in podaci.columns],
})
print('Ukupno redaka:', len(podaci), '| ukupno stupaca:', podaci.shape[1])
print('Praznih ćelija u cijeloj tablici:', int(podaci.isna().sum().sum()))
pregled
```

Vidiš li uzorak? Gotovo sve je zapisano kao **tekst** (`object`), čak i dob. Google obrasci ne šalju tipove — šalju ono što je čovjek upisao. Zato je sljedeći korak priprema podataka: pretvoriti ono što se može pretvoriti u broj, a višestruke odgovore rastaviti na pojedinačne.

Uoči i stupac bez imena (prazan naslov): to je pitanje u obrascu koje nema tekst pitanja, pa mu Google nije dao naslov, a `pandas` ga zove `Unnamed: …`. Takve stvari u stvarnim podacima nisu iznimka nego pravilo.


## 5. Priprema podataka (čišćenje)

Priprema ima tri konkretna zadatka.

**Prvi: dati stupcima imena s kojima se može raditi.** Imena iz obrasca su dugačke rečenice. Preimenovat ćemo ih u kratke nazive (`mjesta`, `dob`, `glazba`, `narodnjaci`, `bach`, `video`) i to tako da ih tražimo po ključnoj riječi — pa kod nastavi raditi i ako predavač kasnije promijeni tekst pitanja.

**Drugi: dob pretvoriti u broj.** Ako u polje za dob netko upiše „dvadeset", `pandas` to ne može pretvoriti; takav zapis postaje prazan i **isključujemo ga iz izračuna sredine** — nikako ne postavljamo nulu, jer bi nula („nula godina") pokvarila sredinu.

**Treći: višestruke odgovore rastaviti na popis.** Kod pitanja s kvačicama u jednoj ćeliji stoji `Kino Art-kino, Klub Crkva, Kafe Dnevni boravak`. Takav zapis ne smijemo brojiti kao jedan odgovor — rastavljamo ga na listu i tek onda brojimo.

Priprema uvijek mora **reći što je promijenila i koliko je odbacila**. Ako to ne kaže, ne znamo vjerovati rezultatima.


```python
cisto = podaci.copy()

# --- imena stupaca na kratke nazive, trazeno po kljucnoj rijeci ---
karta = {}
parovi = [
    ('mjesta', ('izlazite', 'izlazi', 'mjesta')),
    ('dob', ('dob',)),
    ('glazba', ('glazbu slušate', 'glazbu slusate', 'glazba', 'slušate')),
    ('narodnjaci', ('narodnjak',)),
    ('bach', ('bach',)),
    ('video', ('pogledajte video', 'video')),
]
for kratko, kljucevi in parovi:
    nadjeno = nadji_stupac(cisto, *kljucevi)
    if nadjeno is not None:
        karta[nadjeno] = kratko
cisto = cisto.rename(columns=karta)

# --- ako pitanje bez naslova postoji, zovemo ga 'video_izbor' ---
bez_imena = [c for c in cisto.columns if str(c).startswith('Unnamed') or str(c).strip() == '']
if bez_imena:
    cisto = cisto.rename(columns={bez_imena[0]: 'video_izbor'})
print('Preimenovani stupci:', karta if karta else '(nista)', '| bez imena:', bez_imena)

# --- dob u broj ---
cisto['dob_broj'] = pd.to_numeric(cisto['dob'], errors='coerce') if 'dob' in cisto else np.nan
neispravna_dob = int(cisto['dob_broj'].isna().sum()) if 'dob' in cisto else 0

# --- lista mjesta po osobi ---
def u_listu(vrijednost):
    if pd.isna(vrijednost):
        return []
    return [d.strip() for d in str(vrijednost).split(',') if d.strip()]

cisto['mjesta_lista'] = cisto['mjesta'].apply(u_listu) if 'mjesta' in cisto else [[]] * len(cisto)
cisto['broj_mjesta'] = cisto['mjesta_lista'].apply(len)

# --- likert stupci u brojeve ---
for stupac in ('narodnjaci', 'bach'):
    if stupac in cisto:
        cisto[stupac + '_broj'] = pd.to_numeric(cisto[stupac], errors='coerce')

prije = len(cisto)
cisto = cisto[cisto['broj_mjesta'] >= 1].reset_index(drop=True)

print('Redaka prije pripreme:', prije, '| nakon:', len(cisto))
print('Neispravno upisana dob (iskljucena iz izracuna sredine):', neispravna_dob)
if 'narodnjaci_broj' in cisto:
    print('Praznine na skali o narodnjacima:', int(cisto['narodnjaci_broj'].isna().sum()),
          '| na skali o Bachu:', int(cisto['bach_broj'].isna().sum()))
cisto[['mjesta', 'dob', 'glazba']].head(3)
```

## 6. Analiza 1 — gdje se izlazi: frekvencije

Prva analiza u svakom istraživanju je **frekvencijska tablica**: koliko je puta odabrana svaka od osam mogućnosti. Uz broj uvijek gledamo i **udio** (postotak od broja ispitanika), jer bez toga broj ne govori ništa: je li 12 posjeta Kina Art-kino puno? Ako je ispitanika 38, to je trećina svih — puno. Ako ih je 400, to je tri posto — malo.

Dvije stvari na koje pazimo kod ovakvog pitanja:

- **Zbroj postotaka ne mora biti 100 %** (i ne treba biti). Svatko je mogao odabrati dva do pet mjesta, pa jedan ispitanik ulazi u više stupaca. Postoci se računaju **po ispitanicima**, a ne po odgovorima.
- **Najveći broj nije najvažniji nalaz.** Ponekad je zanimljivije ono što je rijetko (kazalište, klasična glazba) — to je mjesto gdje se kulturna publika razlikuje od opće.


```python
svi_odabiri = [m for lst in cisto['mjesta_lista'] for m in lst]
brojevi = pd.Series(Counter(svi_odabiri)).sort_values(ascending=False)
ukupno_osoba = len(cisto)

tablica(brojevi, 'Mjesto', 'Odabira', ukupno=ukupno_osoba).assign(
    **{'udio od ispitanika (%)': [udio(v, ukupno_osoba) for v in brojevi.values]})
```

```python
boje = [CYAN, EMERALD, VIOLET, AMBER, ROSE]
fig, ax = plt.subplots(figsize=(9.5, 4.9))
y = np.arange(len(brojevi))[::-1]
ax.barh(y, brojevi.values, color=[boje[i % len(boje)] for i in range(len(brojevi))], height=0.62)
ax.set_yticks(y)
ax.set_yticklabels(brojevi.index, fontfamily='monospace', fontsize=10)
for yi, vrijednost in zip(y, brojevi.values):
    ax.text(vrijednost + 0.25, yi, str(vrijednost) + '  (' + str(udio(vrijednost, ukupno_osoba)) + ' %)',
            va='center', color='#e2e8f0', fontsize=9.5, fontfamily='monospace')
ax.set_xlabel('broj ispitanika koji su odabrali mjesto (od ' + str(ukupno_osoba) + ')')
ax.set_xlim(0, max(brojevi.values) * 1.22)
ax.grid(axis='x', alpha=0.25)
ax.set_title('Gdje se izlazi u Rijeci', color='white', fontsize=15, fontweight='bold')
plt.tight_layout(rect=[0, 0.07, 1, 1])
fig.text(0.012, 0.02, 'Jedan ispitanik mogao je odabrati više mjesta — zato se postoci ne zbrajaju u 100 %.',
         color='#64748b', fontsize=9, fontfamily='monospace')
plt.show()
```

**Što ova slika govori.** Vidi se osnovna podjela kulturnog izlaska u gradu: kafići i klubovi (svakodnevno, jeftino, društveno) nasuprot institucijama (kazalište, kina — rjeđe, „planski"). Frekvencija sama po sebi ne objašnjava zašto; ona opisuje stanje i daje podlogu za sljedeće pitanje.

**Povezanost: idu li oni koji idu u klub i u kino?** Sljedeći korak je matrica susreta (ko-okurencija) i Jaccardov indeks — mjera preklapanja dvaju skupova od 0 (nitko ne ide u oba) do 1 (svatko kome je jedno važno ide i u drugo).


```python
mjesta = sorted(brojevi.index)
skupovi = [set(lst) for lst in cisto['mjesta_lista']]
pojavnost = {m: sum(1 for s in skupovi if m in s) for m in mjesta}

zajedno = np.zeros((len(mjesta), len(mjesta)), dtype=int)
for s in skupovi:
    for i, a in enumerate(mjesta):
        for j, b in enumerate(mjesta):
            if a in s and b in s:
                zajedno[i, j] += 1

jaccard = np.zeros_like(zajedno, dtype=float)
for i, a in enumerate(mjesta):
    for j, b in enumerate(mjesta):
        unija = pojavnost[a] + pojavnost[b] - zajedno[i, j]
        jaccard[i, j] = zajedno[i, j] / unija if unija else 0.0

parovi = [(jaccard[i, j], mjesta[i], mjesta[j]) for i in range(len(mjesta)) for j in range(i + 1, len(mjesta))]
parovi.sort(reverse=True)
print('Najvece preklapanje (Jaccard):')
for v, a, b in parovi[:5]:
    print('  {:.2f}  {} + {}'.format(v, a, b))

fig, ax = plt.subplots(figsize=(8.4, 6.6))
ax.imshow(jaccard, cmap='viridis', vmin=0, vmax=1)
ax.set_xticks(range(len(mjesta)))
ax.set_xticklabels([m.split(' (')[0] for m in mjesta], rotation=45, ha='right', fontsize=8.5, fontfamily='monospace')
ax.set_yticks(range(len(mjesta)))
ax.set_yticklabels([m.split(' (')[0] for m in mjesta], fontsize=8.5, fontfamily='monospace')
for i in range(len(mjesta)):
    for j in range(len(mjesta)):
        ax.text(j, i, str(round(jaccard[i, j], 2))[1:] if i != j else '—', ha='center', va='center',
                color='#0f172a' if jaccard[i, j] > 0.5 else '#cbd5e1', fontsize=8)
ax.set_title('Koliko se posjeti preklapaju (Jaccardov indeks)', color='white', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 7. Analiza 2 — dob: srednja vrijednost, medijan i raspršenje

Numerički stupac opisujemo s nekoliko brojeva: **srednja vrijednost** (aritmetička sredina), **medijan** (vrijednost u sredini poretka), **standardna devijacija** (koliko su vrijednosti raspršene) te minimum i maksimum.

Dvije lekcije koje se ovdje uvijek nauče:

- **Sredina je osjetljiva na ekstremne vrijednosti, medijan nije.** Ako netko upiše 199 umjesto 19, sredina skoči, a medijan ostane gotovo isti. Zato uz sredinu uvijek gledamo i medijan.

Dob ovdje nije glavni predmet istraživanja — ona je **kontrolna varijabla**: zanima nas je li razlika u ukusu povezana s godinama ili je prisutna u cijeloj skupini.

## 8. Analiza 3 — stavovi na skali 1–5

Dvije tvrdnje na skali slaganja (1 = uopće se ne slažem, 5 = u potpunosti se slažem):

2. *Slušam u slobodno vrijeme glazbu Jochana Sebastiana Bacha.*

Za razliku od dobi, ovdje broj **nije mjera nego oznaka stupnja slaganja** — zato vrijedi pravilo: sredinu smijemo izračunati i koristiti za grubu usporedbu, ali je uvijek prikazujemo uz **cijelu raspodjelu** (koliko je ljudi odabralo 1, 2, 3, 4, 5). Dvije tvrdnje mogu imati istu sredinu, a potpuno različit oblik: jedna može značiti „svi su neutralni", a druga „pola ne, pola da".

Zatim nas zanima **povezanost**: postoji li veza između slaganja s narodnjacima i slaganja s Bachom? Računamo **Spearmanov koeficijent korelacije** (radi s poretkom, ne zahtijeva normalnu raspodjelu):

- +1 znači savršenu istovjetnost poretka, −1 savršenu suprotnost, 0 nema veze;
- vrijednost oko ±0,2 na uzorku od tridesetak ljudi **nije** dokaz veze, nego slaba naznaka;
- i najvažnije: **korelacija nije uzročnost.** Ako su dvije stvari povezane, to još ne znači da jedna uzrokuje drugu — možda obje imaju isti treći uzrok.


```python
prisutne = [s for s in ('narodnjaci_broj', 'bach_broj') if s in cisto]
nazivi = {'narodnjaci_broj': 'narodnjaci u kafićima', 'bach_broj': 'Bach u slobodno vrijeme'}

raspodjele = {}
for s in prisutne:
    serija = cisto[s].dropna().astype(int)
    raspodjele[s] = serija.value_counts().reindex([1, 2, 3, 4, 5]).fillna(0).astype(int)
    print(nazivi[s], '-> sredina', round(serija.mean(), 2), '| medijan', int(serija.median()),
          '| SD', round(serija.std(), 2), '| n =', len(serija))

fig, ax = plt.subplots(figsize=(10.5, 3.8))
boje_skale = ['#fb7185', '#fbbf24', '#a78bfa', '#34d399', '#22d3ee']
lijevo = np.zeros(len(prisutne))
for ocjena in [1, 2, 3, 4, 5]:
    vrijednosti = [raspodjele[s].get(ocjena, 0) for s in prisutne]
    ax.barh([nazivi[s] for s in prisutne], vrijednosti, left=lijevo, color=boje_skale[ocjena - 1],
            label=str(ocjena), height=0.55)
    for i, v in enumerate(vrijednosti):
        if v:
            ax.text(lijevo[i] + v / 2, i, str(int(v)), ha='center', va='center',
                    color='#0f172a', fontsize=9.5, fontweight='bold')
    lijevo = lijevo + np.array(vrijednosti)
ax.set_xlabel('broj studenata')
ax.set_title('Slaganje s tvrdnjama (1 = ne slažem se, 5 = slažem se)', color='white', fontsize=13, fontweight='bold')
ax.legend(title='ocjena', frameon=False, fontsize=9, ncol=5, loc='upper center', bbox_to_anchor=(0.5, -0.28))
plt.tight_layout()
plt.show()
```

```python
# --- povezanost dviju skala (Spearman) ---
if len(prisutne) == 2:
    par = cisto[prisutne].dropna()
    from scipy import stats
    rho, p = stats.spearmanr(par[prisutne[0]], par[prisutne[1]])
    print('Zajednicki ispitanici:', len(par))
    print('Spearmanov rho = {:.2f} | p = {:.3f}'.format(rho, p))
    if abs(rho) < 0.2:
        print('Tumacenje: veza je slaba ili je nema — ukusi se ne krecu u istom smjeru.')
    elif abs(rho) < 0.5:
        print('Tumacenje: umjerena veza; ali s ovako malim uzorkom ne bismo tvrdili uzrocnost.')
    else:
        print('Tumacenje: izrazena veza u poretku odgovora; provjeriti na vecem uzorku.')

    rnd = np.random.default_rng(7)
    x = par[prisutne[0]].astype(float) + rnd.uniform(-0.13, 0.13, len(par))
    y = par[prisutne[1]].astype(float) + rnd.uniform(-0.13, 0.13, len(par))
    fig, ax = plt.subplots(figsize=(6.6, 5.2))
    ax.scatter(x, y, s=70, color=CYAN, alpha=0.75, edgecolor=PDV)
    ax.set_xticks([1, 2, 3, 4, 5])
    ax.set_yticks([1, 2, 3, 4, 5])
    ax.set_xlabel('narodnjaci u kafićima')
    ax.set_ylabel('Bach u slobodno vrijeme')
    ax.set_title('Odgovor po odgovoru (male nasumične pomake radi preglednosti)',
                 color='white', fontsize=11.5, fontweight='bold')
    ax.grid(alpha=0.2)
    plt.tight_layout()
    plt.show()
```

## 9. Analiza 4 — glazba: slobodni unos

Polje „Koju glazbu slušate" otvoreno je pitanje, i tu se vidi prava narav podataka iz ankete. Studenti su pisali ono što im padne na pamet:

- razdvajali su izvođače **točkom sa zarezom** (kako je traženo), ali i zarezom, kosom crtom ili novim redom;
- pisali su **velikim i malim slovima** („the beatles", „The Beatles"), ponekad s razmakom prije zareza;
- navodili su **žanr** umjesto izvođača („klasična glazba"), **sastave** („Florence + The Machine") i **imena s dijakriticima** („Riblja Čorba").

Prije brojanja radimo tri poteza: rastavi na pojedinačne izvođače, skini razmake i sve svedi na mala slova, pa tek onda broji. Uz to vodimo **zapis o tome koliko se sirovih inačica spojilo** — to je najbolja ilustracija zašto je priprema podataka 60 % posla.

Kad brojanje bude gotovo, pogledat ćemo i **koliko su ukusi međusobno slični** — prvo klasičnom vektorskom metodom (TF-IDF), a zatim, ako imamo ključ, i semantičkim embeddingom.


```python
def izvodaci(tekst):
    if pd.isna(tekst):
        return []
    dijelovi = re.split(r';|,|/|\|| i | and ', str(tekst))
    return [d.strip().strip('.').lower() for d in dijelovi if len(d.strip()) > 1]

cisto['izvodaci'] = cisto['glazba'].apply(izvodaci)
svi = [a for lst in cisto['izvodaci'] for a in lst]
brojanje = pd.Series(Counter(svi)).sort_values(ascending=False)

sirovo = Counter()
for tekst in cisto['glazba'].dropna():
    for d in re.split(r';|,|/|\|', str(tekst)):
        if len(d.strip()) > 1:
            sirovo[d.strip()] += 1
print('Razlicitih zapisa prije ciscenja:', len(sirovo), '-> nakon ciscenja:', len(brojanje))
print('Ukupno navedenih izvodaca (svih ispitanika):', len(svi))

tablica(brojanje.head(15), 'Izvođač', 'Navoda', ukupno=len(cisto)).assign(
    **{'udio ispitanika (%)': [udio(v, len(cisto)) for v in brojanje.head(15).values]})
```

```python
vrh = brojanje.head(15)[::-1]
fig, ax = plt.subplots(figsize=(9.5, 5.9))
ax.barh(vrh.index, vrh.values, color=[VIOLET if i % 2 else CYAN for i in range(len(vrh))], height=0.65)
for i, v in enumerate(vrh.values):
    ax.text(v + 0.15, i, str(int(v)), va='center', color='#e2e8f0', fontsize=9, fontfamily='monospace')
ax.set_xlabel('broj navoda (isti izvođač kod više studenata)')
ax.set_title('Najčešće navedeni izvođači', color='white', fontsize=14, fontweight='bold')
plt.tight_layout(rect=[0, 0.06, 1, 1])
fig.text(0.012, 0.018, 'Imena su svedena na mala slova i razdvojena po ; , / | kako bi se ista imena spojila.',
         color='#64748b', fontsize=8.6, fontfamily='monospace')
plt.show()

print('Presjek ukusa: izvodaca koje navodi samo jedna osoba:',
      sum(1 for a, n in brojanje.items() if n == 1), 'od', len(brojanje))
```

```python
RUCNA_MAPA = {
    'the beatles': 'rock', 'beatles': 'rock', 'radiohead': 'rock', 'arctic monkeys': 'rock',
    'the strokes': 'rock', 'fontaines d.c.': 'rock', 'pink floyd': 'rock', 'led zeppelin': 'rock',
    'deep purple': 'rock', 'ac/dc': 'rock', 'metallica': 'metal', 'iron maiden': 'metal',
    'rammstein': 'metal', 'system of a down': 'metal', 'queen': 'rock', 'david bowie': 'rock',
    'taylor swift': 'pop', 'billie eilish': 'pop', 'lana del rey': 'pop', 'adele': 'pop',
    'ed sheeran': 'pop', 'coldplay': 'pop', 'olivia rodrigo': 'pop', 'lorde': 'pop',
    'kendrick lamar': 'hip-hop', 'travis scott': 'hip-hop', 'drake': 'hip-hop',
    'the weeknd': 'r&b', 'sza': 'r&b', 'doja cat': 'pop', 'frank ocean': 'r&b',
    'bach': 'klasična', 'chopin': 'klasična', 'vivaldi': 'klasična', 'mozart': 'klasična',
    'beethoven': 'klasična', 'schubert': 'klasična', 'brahms': 'klasična', 'klasična glazba': 'klasična',
    'miles davis': 'jazz', 'john coltrane': 'jazz', 'nina simone': 'jazz', 'etta james': 'jazz',
    'bts': 'k-pop', 'blackpink': 'k-pop', 'newjeans': 'k-pop',
    'severina': 'pop/folk', 'jelena rozga': 'pop/folk', 'thompson': 'pop/folk', 'nina badrić': 'pop/folk',
    'gibonni': 'pop/folk', 'dino merlin': 'pop/folk', 'zdravko čolić': 'pop/folk',
    'bijelo dugme': 'rock', 'bajaga': 'rock', 'riblja čorba': 'rock', 'idoli': 'rock',
    'bob dylan': 'folk', 'leonard cohen': 'folk', 'hozier': 'folk',
    'dubioza kolektiv': 'rock', 'zabranjeno pušenje': 'rock', 'hladno pivo': 'rock',
    'let 3': 'rock', 'tbf': 'rock', 'elemental': 'hip-hop', 'tram 11': 'hip-hop',
    'nipplepeople': 'elektronika', 'svemirko': 'indie', 'porto morto': 'indie',
    'florence + the machine': 'indie', 'josipa lisac': 'pop/folk', 'vlatko stefanovski': 'jazz',
    'gracie abrams': 'pop', 'tyler, the creator': 'hip-hop', 'prince': 'pop', 'alanis morissette': 'rock',
    'širom svijeta': 'rock',
}

ZANROVI = ['rock', 'pop', 'metal', 'hip-hop', 'r&b', 'jazz', 'klasična', 'folk',
           'elektronika', 'indie', 'k-pop', 'pop/folk', 'ostalo']


def zanr_preko_modela(izvodjac, keš={}):
    'Pita model za zanr; pri prvom pozivu postavlja ugovor izlaza. Vraca (zanr, izvor).'
    if izvodjac in keš:
        return keš[izvodjac], 'keš'
    zadatak = ('Ti si glazbeni analiticar. Za izvodaca navedi glavni glazbeni zanr.\n'
               'Vrati ISKLJUCIVO JSON oblika {"izvodjac": "...", "zanr": "..."}.\n'
               'Dopusteni zanrovi: ' + ', '.join(ZANROVI) + '.\n'
               'Ako izvodaca ne poznajes, vrati "nepoznato". Ne dodaj nikakvo objasnjenje.\n'
               'Izvodac: ' + izvodjac)
    for pokusaj in range(3):
        try:
            odgovor = klijent.models.generate_content(
                model='gemini-3.5-flash', contents=zadatak,
                config={'response_mime_type': 'application/json', 'temperature': 0.0})
            podatak = json.loads(odgovor.text)
            zanr = str(podatak.get('zanr', 'nepoznato')).strip().lower()
            if zanr in ZANROVI:
                keš[izvodjac] = zanr
                return zanr, 'model'
            break
        except Exception as greska:
            if '429' in str(greska) or 'quota' in str(greska).lower():
                time.sleep(5)
                continue
            print('preskacem', izvodjac, '->', type(greska).__name__, str(greska)[:70])
            break
    zanr = RUCNA_MAPA.get(izvodjac, 'nepoznato')
    keš[izvodjac] = zanr
    return zanr, 'rucni popis'


if RADI_AI:
    izvori_zanra = {}
    for a in list(brojanje.head(20).index):
        zanr, izvor = zanr_preko_modela(a)
        izvori_zanra[a] = (zanr, izvor)
    print('Kategorizirano izvodaca:', len(izvori_zanra))
else:
    izvori_zanra = {a: (RUCNA_MAPA.get(a, 'nepoznato'), 'rucni popis') for a in list(brojanje.head(20).index)}
    print('Model nije pozvan (nema kljuca) — koristen je rucni popis zanrova za pokazne podatke.')

popis = pd.DataFrame([(a, z, i) for a, (z, i) in izvori_zanra.items()],
                     columns=['izvodac', 'zanr', 'izvor']).sort_values('izvodac')
print(popis.to_string(index=False))
```

```python
# Zanr po ispitaniku: svakom studentu pridruzujemo zanrove njegovih izvodaca
def zanrovi_osobe(lista):
    return [izvori_zanra.get(a, (RUCNA_MAPA.get(a, 'nepoznato'), 'rucni popis'))[0] for a in lista]

cisto['zanrovi'] = cisto['izvodaci'].apply(zanrovi_osobe)
brojanje_zanrova = pd.Series(Counter([z for lst in cisto['zanrovi'] for z in lst if z != 'nepoznato']))
brojanje_zanrova = brojanje_zanrova.sort_values(ascending=False)
nepoznato = sum(1 for lst in cisto['zanrovi'] for z in lst if z == 'nepoznato')
print('Zanrova prepoznato:', int(brojanje_zanrova.sum()), '| ostalo nepoznato:', nepoznato)

fig, ax = plt.subplots(figsize=(9.2, 4.4))
ax.bar(brojanje_zanrova.index, brojanje_zanrova.values,
       color=[CYAN, EMERALD, VIOLET, AMBER, ROSE] * 3, alpha=0.9)
for i, v in enumerate(brojanje_zanrova.values):
    ax.text(i, v + 0.2, str(int(v)), ha='center', color='#e2e8f0', fontsize=9.5, fontfamily='monospace')
plt.xticks(rotation=35, ha='right', fontfamily='monospace', fontsize=9)
ax.set_ylabel('broj navoda')
ax.set_title('Žanrovi u glazbenom ukusu skupine', color='white', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

### 10.2. Sličnost ukusa: klasična vektorska metoda i embedding

Sada nas zanima **tko je kome blizak po ukusu**. Dvije metode, i zgodno ih je usporediti:

**TF-IDF (klasična metoda, radi bez modela).** Svakog ispitanika pretvaramo u vektor prema slovima i riječima koje se pojavljuju u njegovu popisu izvođača. Rijetke riječi dobivaju veću težinu (zato *inverse document frequency*). Sličnost mjerimo **kosinusom** između vektora. Prednost: radi uvijek, besplatno. Nedostatak: „AC/DC" i „Metallica" nisu slični jer nemaju zajedničkih slova, iako su isti žanr.

**Embedding (semantička metoda, treba ključ).** Model pretvara tekst u vektor od nekoliko stotina brojeva koji opisuju **značenje**. Tu su „AC/DC" i „Metallica" blizu. Prednost: hvata značenje. Nedostatak: treba ključ, kvotu i vrijeme.

Usporedba dviju matrica sličnosti (koliko se slažu u poretku parova) lijepa je ilustracija koliko nam semantika donosi — i kad se isplati platiti cijenu.


```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.cluster import KMeans

tekstovi = cisto['glazba'].fillna('').apply(lambda t: ' ; '.join(izvodaci(t)))
vektorizator = TfidfVectorizer(analyzer='char_wb', ngram_range=(2, 4), min_df=1)
X = vektorizator.fit_transform(tekstovi)
sl_tfidf = cosine_similarity(X)
print('Matrica slicnosti (TF-IDF):', sl_tfidf.shape)

grupe = KMeans(n_clusters=3, n_init=10, random_state=0).fit_predict(X)
cisto['grupa_tfidf'] = grupe
for g in sorted(set(grupe)):
    pod = brojanje.loc[[a for a in brojanje.index if a in set(sum(cisto.loc[cisto['grupa_tfidf'] == g, 'izvodaci'], []))]]
    print('Grupa', g, '(n =', int((grupe == g).sum()), ') najcesci izvodaci:',
          ', '.join(list(pod.head(6).index)))

poredak = np.argsort(grupe)
sl_sredeno = sl_tfidf[np.ix_(poredak, poredak)]
fig, ax = plt.subplots(figsize=(7.4, 6.2))
ax.imshow(sl_sredeno, cmap='magma', vmin=0, vmax=1)
granice = np.cumsum([int((grupe == g).sum()) for g in sorted(set(grupe))])[:-1]
for b in granice:
    ax.axhline(b - 0.5, color=CYAN, linewidth=1.2)
    ax.axvline(b - 0.5, color=CYAN, linewidth=1.2)
ax.set_title('Slicnost ukusa (TF-IDF): ispitanici poredani po grupama',
             color='white', fontsize=13, fontweight='bold')
ax.set_xlabel('ispitanik'); ax.set_ylabel('ispitanik')
plt.tight_layout()
plt.show()
```

```python
# Semanticka slicnost preko embeddinga — samo ako imamo kljuc
uskladenost = None
if RADI_AI:
    try:
        from scipy import stats
        odgovor = klijent.models.embed_content(model='gemini-embedding-2', contents=list(tekstovi))
        E = np.array([e.values for e in odgovor.embeddings])
        E = E / np.linalg.norm(E, axis=1, keepdims=True)
        sl_emb = E @ E.T
        gornji = np.triu_indices_from(sl_tfidf, k=1)
        uskladenost, _ = stats.spearmanr(sl_tfidf[gornji], sl_emb[gornji])
        print('Slicnost matrica (Spearman izmedju TF-IDF i embeddinga): {:.2f}'.format(uskladenost))
        print('Broj dimenzija embeddinga:', E.shape[1])
        poredak2 = np.argsort(grupe)
        fig, ax = plt.subplots(figsize=(7.4, 6.2))
        ax.imshow(sl_emb[np.ix_(poredak2, poredak2)], cmap='magma', vmin=np.percentile(sl_emb, 5), vmax=1)
        for b in granice:
            ax.axhline(b - 0.5, color=EMERALD, linewidth=1.2)
            ax.axvline(b - 0.5, color=EMERALD, linewidth=1.2)
        ax.set_title('Slicnost ukusa (embedding): iste grupe, semanticka blizina',
                     color='white', fontsize=13, fontweight='bold')
        ax.set_xlabel('ispitanik'); ax.set_ylabel('ispitanik')
        plt.tight_layout()
        plt.show()
        if uskladenost is not None and uskladenost < 0.5:
            print('Matrice se razlikuju: embedding vidi blizinu i tamo gdje nema zajednickih rijeci.')
    except Exception as greska:
        print('Embedding nije uspio:', type(greska).__name__, str(greska)[:120])
else:
    print('Bez kljuca nema embeddinga — ostajemo na TF-IDF matrici iz prethodne celije.')
    print('Usporedba dviju metoda je odlicna vjezba za domacu zadacu (vidi zadatke na kraju).')
```

### 10.3. Sažetak nalaza — model piše iz naših brojki

Završni korak je tekst. Model dobiva **isključivo izračunate brojke** koje smo upravo proizveli i zadatak da napiše pet do sedam rečenica nalaza — bez novih podataka, bez nagađanja, bez savjeta koje nitko nije tražio.

Ovo je i test za nas: ako model u sažetku spomene brojku koje nema u našem popisu, to je halucinacija i takav tekst ne ide u rad. Provjera je jednostavna — svaka brojka u tekstu mora postojati u brojkama koje smo mu dali.


```python
sazetak_brojeva = {
    'broj ispitanika': int(len(cisto)),
    'odaziv na pitanje o mjestima': int(cisto['broj_mjesta'].gt(0).sum()),
    'tri najodabranija mjesta': {str(k): int(v) for k, v in list(brojevi.head(3).items())},
    'najrjeđe odabrano mjesto': {str(brojevi.index[-1]): int(brojevi.values[-1])},
    'srednja dob': round(float(dob.mean()), 2),
    'medijan dobi': float(dob.median()),
    'najveća povezanost mjesta (Jaccard)': {'par': [parovi[0][1], parovi[0][2]], 'vrijednost': round(parovi[0][0], 2)},
    'sredina slaganja s narodnjacima': round(float(cisto['narodnjaci_broj'].mean()), 2) if 'narodnjaci_broj' in cisto else None,
    'sredina slaganja s Bachom': round(float(cisto['bach_broj'].mean()), 2) if 'bach_broj' in cisto else None,
    'pet najcešćih izvođača': {str(k): int(v) for k, v in list(brojanje.head(5).items())},
    'najcešći žanrovi': {str(k): int(v) for k, v in list(brojanje_zanrova.head(5).items())},
}

print(json.dumps(sazetak_brojeva, ensure_ascii=False, indent=1))

if RADI_AI:
    zadatak = ('Ti si istrazivac u drustvenim znanostima. Napisi sazetak nalaza (pet do sedam recenica) '
               'ISKLJUCIVO na temelju brojki iz prilozenog JSON-a. Pravila: ne izmisljaj nijednu brojku '
               'koje nema u JSON-u; ne tvrdi uzrocnost; na kraju navedi jedno ogranicenje istrazivanja.\n'
               'Brojke: ' + json.dumps(sazetak_brojeva, ensure_ascii=False))
    try:
        odgovor = klijent.models.generate_content(model='gemini-3.5-flash', contents=zadatak,
                                                  config={'temperature': 0.2})
        print('\n--- SAZETAK (napisao model iz nasih brojki) ---\n' + odgovor.text)
    except Exception as greska:
        print('Model nije odgovorio:', type(greska).__name__, str(greska)[:120])
else:
    print('\n--- SAZETAK (sastavljen iz brojki, bez modela) ---')
    print('Prikupljeno je ' + str(len(cisto)) + ' odgovora. Najcesce se izlazi u: ' +
          ', '.join([str(k) for k in brojevi.head(3).index]) + '. Srednja dob ispitanika je ' +
          str(round(float(dob.mean()), 1)) + ' godine, a medijan ' + str(int(dob.median())) +
          '. Najvece preklapanje pokazuju ' + parovi[0][1] + ' i ' + parovi[0][2] + '. '
          'Medju izvodacima prednjace ' + ', '.join(list(brojanje.head(3).index)) + '. '
          'Ogranicenje: uzorak je mali i prikupljen u jednoj skupini, pa zakljucci vrijede za ovu skupinu.')
```

## 12. Što smo naučili i gdje dalje

Prošli smo cijeli put od obrasca do podatkovne priče. Sažetak:

- **Podaci iz obrasca su neuredni po prirodi** — više odgovora u jednoj ćeliji, tekst umjesto broja, različita pisanja istog imena. Priprema nije dosadan uvod nego dio metode.
